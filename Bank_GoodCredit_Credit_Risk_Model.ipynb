# 4. Exploratory Data Analysis – Findings and Decisions

## 4.1 Dataset Overview

The project uses three datasets containing historical account information, customer enquiries and demographic/application information.

| Dataset | Records | Columns | Unique Customers |
|---|---:|---:|---:|

After customer-level aggregation and integration, the final modelling dataset contained **23,896 customers and 96 columns**.

**Decision:** Account and enquiry records were aggregated at customer level before merging with the demographics dataset. This preserved historical information while creating one modelling record per customer.

## 4.2 Target Variable Distribution

The target variable `Bad_label` has the following distribution:

| Target | Description | Customers | Percentage |
|---|---|---:|---:|
| 0 | Good credit history | 22,892 | 95.80% |
| 1 | Bad credit history (30 DPD+) | 1,004 | 4.20% |
| Total | | 23,896 | 100% |

Only 4.20% of customers belong to the bad-credit category, indicating a significant class imbalance.

**Decision:** Stratified train-test splitting was used to preserve the target distribution. ROC-AUC and Gini were prioritised for evaluating the model's ability to rank customers by credit risk.

## 4.3 Missing Value Analysis

Significant missing values were identified in the account dataset.

| Feature | Missing Percentage |
|---|---:|
| amt_past_due | 99.53% |
| rateofinterest | 86.67% |
| cashlimit | 81.02% |
| actualpaymentamount | 77.93% |
| creditlimit | 73.74% |
| closed_dt | 58.60% |
| last_paymt_dt | 13.64% |
| high_credit_amt | 4.74% |

The enquiry dataset had only approximately 0.03% missing values in enquiry-related fields. The demographics dataset had 15 missing `entry_time` values.

**Decisions:**
- Missing values were not treated as zero automatically.
- Numerical and categorical features were handled through appropriate preprocessing.
- Median imputation was used for numerical features.
- Most-frequent imputation was used for categorical features.
- Highly incomplete fields, particularly `amt_past_due`, were treated cautiously.

## 4.4 Duplicate Record Analysis

Initially, the source account and enquiry datasets contained exact duplicate records:

- Cust_Account: 2,401 duplicate rows.
- Cust_Enquiry: 13,827 duplicate rows.
- Cust_Demographics: No duplicate rows.

The exact duplicate rows were removed from the account and enquiry datasets.

Repeated customer IDs in account and enquiry tables were retained because customers can have multiple historical accounts and enquiries.

After cleaning and integration, the final dataset contained 23,896 unique customers without duplicate customer IDs.

**Decision:** Exact duplicate records were removed, while legitimate multiple account and enquiry records were retained for customer-level aggregation.

## 4.5 Numerical Feature Analysis

The account dataset contained highly variable financial values.

Key observations included:

- Current account balance had a median of zero.
- 8,781 account records contained negative current balances.
- Credit limits and actual payment amounts had substantial missingness.
- High-credit amounts and current balances showed large ranges.
- Past-due amounts were available for only 865 account records.

The enquiry amount distribution was highly right-skewed:

| Statistic | Enquiry Amount |
|---|---:|
| Mean | 823,110 |
| Median | 50,000 |
| 75th Percentile | 121,000 |
| Maximum | 999,999,999 |

The mean being substantially higher than the median indicates the presence of extreme enquiry amounts.

**Decisions:**
- Numerical fields were converted into appropriate numeric formats.
- Extreme values were investigated rather than automatically deleted.
- Negative balances were retained because their meaning was not established as invalid.
- Highly skewed financial variables were handled cautiously.

## 4.6 Categorical Feature Analysis

The demographic dataset contained anonymised categorical features with varying levels of cardinality.

Some features had particularly high numbers of unique categories:

- feature_47: 11,250
- feature_20: 10,610
- feature_21: 8,463
- feature_38: 5,664
- feature_45: 4,473

The enquiry purpose field contained 37 unique codes. Purpose code 10 accounted for approximately 58.38% of enquiry records.

**Decisions:**
- High-cardinality features were inspected before preprocessing.
- Selected identifier-like and masked features were excluded from the modelling inputs.
- Categorical encoding was performed using OneHotEncoder with infrequent-category grouping.
- Enquiry purpose codes were retained without assigning unverified business meanings.

## 4.7 Date Feature Analysis

Date fields were converted into datetime format for analysis.

The customer application reference date, `dt_opened`, ranged from 16 April 2015 to 31 December 2015.

Account and enquiry dates were examined against customer reference dates. No future reporting or enquiry dates were identified in the earlier validation.

**Decisions:**
- Date fields were converted into datetime format.
- The customer application date was used as the reference date for enquiry recency calculations.
- Date differences were used where appropriate for feature engineering.
- Missing dates were handled without assuming an arbitrary replacement date.

## 4.8 Payment History Analysis

The account dataset contained two payment-history fields:

- `paymenthistory1`
- `paymenthistory2`

The fields contained codes such as `STD`, `XXX`, `000`, `030`, `060`, `090` and `120`.

Payment-history strings had varying lengths. In particular, `paymenthistory2` contained 106,395 empty records.

**Decision:** Payment-history codes were not decoded without a verified codebook. This avoided making unsupported assumptions about the meaning and sequence of repayment statuses.

```python
# Create customer-level enquiry features

# Reference date for each customer
reference_dates = demographics[
    ['customer_no', 'dt_opened']
]

# Attach customer reference dates to enquiry records
enquiry_temp = enquiry.merge(
    reference_dates,
    on='customer_no',
    how='left',
    suffixes=('_enquiry', '_customer')
)

# Calculate days between enquiry and customer reference date
enquiry_temp['days_before_application'] = (
    enquiry_temp['dt_opened_customer'] -
    enquiry_temp['enquiry_dt']
).dt.days

# Aggregate enquiry features
enquiry_features = enquiry_temp.groupby('customer_no').agg(
    enquiry_count=('customer_no', 'count'),
    avg_enquiry_amount=('enq_amt', 'mean'),
    total_enquiry_amount=('enq_amt', 'sum'),
    count_enquiry_recency_90=(
        'days_before_application',
        lambda x: ((x >= 0) & (x <= 90)).sum()
    ),
    count_enquiry_recency_365=(
        'days_before_application',
        lambda x: ((x >= 0) & (x <= 365)).sum()
    )
).reset_index()

print("Enquiry features shape:", enquiry_features.shape)
enquiry_features.head()
```
